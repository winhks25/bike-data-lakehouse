# Reading from source files and writing into databases

```python
crm_files = ["sales_details", "prd_info", "cust_info"]
erp_files = ["PX_CAT_G1V2", "LOC_A101", "CUST_AZ12"]

for i in range(len(crm_files)):
    df = (
        spark.read.option("header", "true")
        .option("inferSchema", "true")
        .csv("/Volumes/workspace/bronze/source_system/source_crm/" + crm_files[i] + ".csv")
    )
    df.write.mode("overwrite").saveAsTable("workspace.bronze.crm_" + crm_files[i])


for i in range(len(erp_files)):
    df = (
        spark.read.option("header", "true")
        .option("inferSchema", "true")
        .csv("/Volumes/workspace/bronze/source_system/source_erp/" + erp_files[i] + ".csv")
    )
    df.write.mode("overwrite").saveAsTable("workspace.bronze.erp_" + erp_files[i])
```
